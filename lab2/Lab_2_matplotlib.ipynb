```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 67]
y = [2, 4, 6, 94, 222]

plt.plot(x, y)
plt.title("Line plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = [1, 2, 3, 4, 67]
y = [2, 4, 6, 94, 222]

fig, ax = plt.subplots()

ax.plot(x, y, marker="o", label="Data")

ax.set_title("Matplotlib")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")

ax.legend()
plt.show()
```

```python
# Plot the monthly temperature of 2 cites
# Plot a line plot in the same graph

months = [
  "January", "February", "March", "April", "May", "June",
  "July", "August", "September", "October", "November", "December"
];

mangaloreTemp = [
  26.4, 27.1, 28.2, 28.9, 28.0, 26.5,
  25.8, 25.2, 26.0, 26.8, 27.2, 26.8
];

udupiTemp = [
  26.5, 27.0, 28.0, 28.7, 28.2, 26.5,
  25.6, 25.7, 26.0, 26.9, 27.3, 26.9
];
```

```python
plt.figure(figsize=(14, 6))

plt.plot(months, mangaloreTemp, udupiTemp)
```

```python
plt.figure(figsize=(14, 6))

plt.plot(months, mangaloreTemp, label="Mangalore", color="#30EBFF", marker="X")
plt.plot(months, udupiTemp, label="Udupi", color="#FF80F5", marker="v")
plt.title("Monthly temperature of Mangalore v/s Udupi")

plt.xlabel("MONTHS")
plt.ylabel("TEMPERATURE (°C)")

plt.legend()
plt.plot()
```

```python
fig, tx = plt.subplots(figsize=(14,6))

fig

tx.plot(months, mangaloreTemp, label="MANGALORE", color="#30EBFF", marker="X")
tx.plot(months, udupiTemp, label="UDUPI", color="#FF80F5", marker="v")

tx.set_title("Monthly temperature of Mangalore v/s Udupi")
tx.set_xlabel("MONTHS")
tx.set_ylabel("TEMPERATURE (°C)")

tx.legend()
tx.set_facecolor("#000000")
```
