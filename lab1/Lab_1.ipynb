```python
# Assignment

"""
Use the titanic dataset
1. Total number of passengers
2. Find the average age of the survivors and the non survivors
3. List the passengers who are older than 60 years
4. Find the average fare for each passenger class
"""
```

```python
titanic = pd.read_csv("./titanic.csv")

titanic.head()
```

```python
print("Total number os passengers", len(titanic))
```

```python
survivors = titanic[titanic['Survived'] == 1]['Age'].dropna()

print(np.average(survivors))

nonSurvivors = titanic[titanic['Survived'] == 0]['Age'].dropna()

print(np.average(nonSurvivors))
```

```python
titanic[titanic['Age'] > 60].head()
```

```python
passengerClass1 = titanic[titanic['Pclass'] == 1]
print("Average of passenger class 1", np.average(passengerClass1['Fare']))

passengerClass2 = titanic[titanic['Pclass'] == 2]
print("Average of passenger class 2", np.average(passengerClass2['Fare']))

passengerClass3 = titanic[titanic['Pclass'] == 3]
print("Average of passenger class 3", np.average(passengerClass3['Fare']))
```
