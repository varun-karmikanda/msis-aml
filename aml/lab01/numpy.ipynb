```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print('Matrix addition: \n', A + B)
print('Element-wise multiplication: \n', A * B)
print('Matrix multiplication: \n', A.dot(B)) # For 2D, Dot Prod and Matrix Mul is same
print('Transpose of A: \n', A.T)
print('Element-wise multiplication: \n', A @ B)


# explain the difference between:
# A * B and A @ B

# A * B => Elementwise Multiplication
# A @ B => Dot Product: Sum of product in MatrixMultiplication(A, B)
```

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])


print('Mean temperature: ', np.mean(temperature))
print('Maximum and minimum temperatures: ', np.max(temperature), ", ", np.min(temperature))
print('Number of days above the average temperature: ', len(temperature[temperature > np.mean(temperature)]))
print('Temperatures greater than 34°C: ', temperature[temperature > 34])
print('Difference between maximum and minimum: ', np.max(temperature) - np.min(temperature))
print('Standard deviation: ', np.std(temperature))
```

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])



print('Total annual sales: ', np.sum(sales))
print('Average monthly sales: ', np.mean(sales))
print('Month with the highest sales: ', np.argmax(sales) + 1)
print('Month with the lowest sales: ', np.argmin(sales) + 1)
print('Number of months with sales above ₹20,000: ', len(sales[sales > 20000]))
print('Percentage increase from January to December: ', ((sales[-1:] - sales[0]) / sales[0]) * 100)
```

```python
# Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.

import time

n=1000000
py_list=list(range(1,n+1))
np_array = np.arange(1,n+1)

start_time1= time.time()
py_result = [x * 2 for x in py_list]
py_time = time.time() - start_time1

start_time = time.time()
np_result = np_array * 2
np_time = time.time() - start_time

print(f"Python list time: {py_time:.5f} seconds")
print(f"NumPy array time: {np_time:.5f} seconds")
print(f"NumPy is {py_time / np_time:.1f}x faster")
```
