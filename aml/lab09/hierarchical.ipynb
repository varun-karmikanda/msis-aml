```python
import pandas as pd
from pathlib import Path 
```

```python
dataset_path = Path(r"C:\Users\VarunKarmikanda\msis\sem1\aml\aml\data\mall.csv")
dataset = pd.read_csv(dataset_path)

dataset.head()
```

```python
X = dataset.iloc[:, [3, 4]].values
X[: 10]
```

```python
# Dendogram method for optimal clusters
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

linkage_data = linkage(X, method="ward")
dendogram = dendrogram(linkage_data)
# ward for min increase in the within clister variance
plt.title("Dendogram method")
plt.xlabel("Customer")
plt.ylabel("Euclidean distance")
```

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=5, metric="euclidean", linkage="ward")

y_hc = hc.fit_predict(X)
y_hc
```

```python
import numpy as np

plt.scatter(X[y_hc == 0, 0], X[y_hc == 0, 1], label="Careful")
plt.scatter(X[y_hc == 1, 0], X[y_hc == 1, 1], label="Standard")
plt.scatter(X[y_hc == 2, 0], X[y_hc == 2, 1], label="Target")
plt.scatter(X[y_hc == 3, 0], X[y_hc == 3, 1], label="Careless")
plt.scatter(X[y_hc == 4, 0], X[y_hc == 4, 1], label="Sensible")

centroids = np.array([X[y_hc == i].mean(axis=0) for i in range(5)])
centroids
plt.scatter(centroids[:, 0], centroids[:, 1], c="magenta", marker="*", s=150, label="Centroid")

plt.title("Cluster of clients")
plt.xlabel("Customer")
plt.ylabel("Sepending score")
plt.legend()
```
